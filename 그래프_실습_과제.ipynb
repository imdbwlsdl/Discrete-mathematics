```python
graph = {
    'Yujin Kim' : ['Gyuwon Yang', 'Geonwoo Kim', 'Woojin Choi', 'Jieun Lee', 'Gwanghyun Song'],
    'Gyuwon Yang' : ['Woojin Choi', 'Woochan Jung', 'Soeun Jeon', 'Daon Min'],
    'Yeyoung Hong' : ['Yoonseo Kim', 'Seoyeon Hong', 'Hanna Jeon', 'Jiwoo Jung'],
    'Geonwoo Kim' : ['Yujin Kim', 'Junseong Kim', 'Sangmuk Shim', 'Geonwoo Park', 'Dongseong Bae'],
    'Seongwon Lee' : ['Jihyun Lee', 'Sechan Kim', 'Jaeho Shin', 'Seongjun Moon', 'Donghun Shin']
}

def dfs(graph, v, visited):
  visited.add(v)
  print(v, end='')
  for neighbor in graph.get(v, []):
    if neighbor not in visited:
      dfs(graph, neighbor, visited)

def DFS_all(graph):
  visited = set()
  for u in graph:
    if u not in visited:
      dfs(graph, u, visited)

print("DFS 탐색 순서:")
DFS_all(graph)

from collections import deque

def bfs(graph, start, visited):
  queue = deque([start])
  visited.add(start)

  while queue:
    v = queue.popleft()
    print(v, end='')
    for neighbor in graph.get(v, []):
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

def BFS_all(graph):
  visited = set()
  for u in graph:
    if u not in visited:
      bfs(graph, u, visited)

print("\nBFS 탐색 순서: ")
BFS_all(graph)

from collections import deque

def recommend_friend(graph, start):
  visited = set()
  queue = deque([(start, 0)])

  print(f"\n{start}의 친구 추천 (2단계):")

  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 2:
        print("-", current)
      if depth < 2:
        for neighbor in graph.get(current, []):
          queue.append((neighbor, depth + 1))

recommend_friend(graph, 'Geonwoo Kim')
```

```python
from collections import deque  # 큐(줄 서기) 자료구조: 먼저 들어온 사람이 먼저 나감

# ── 1. 그래프 (인접 리스트) ─────────────────────────────
# key: 사람, value: 그 사람과 직접 접촉한 사람 목록
graph = {
    'Yujin Kim' : ['Gyuwon Yang', 'Geonwoo Kim', 'Woojin Choi', 'Jieun Lee', 'Gwanghyun Song'],
    'Gyuwon Yang' : ['Woojin Choi', 'Woochan Jung', 'Soeun Jeon', 'Daon Min'],
    'Yeyoung Hong' : ['Yoonseo Kim', 'Seoyeon Hong', 'Hanna Jeon', 'Jiwoo Jung'],
    'Geonwoo Kim' : ['Yujin Kim', 'Junseong Kim', 'Sangmuk Shim', 'Geonwoo Park', 'Dongseong Bae'],
    'Seongwon Lee' : ['Jihyun Lee', 'Sechan Kim', 'Jaeho Shin', 'Seongjun Moon', 'Donghun Shin']
}

# ── 2. 감염 경로 추적 함수 (BFS) ──────────────────────
def trace_infection(graph, start, max_level=2):
    """
    start(최초 감염자)에서 max_level 단계 이내로 전염되는 사람을 찾는다.
    반환값: {사람: (레벨, 감염 경로 리스트)}  ※ 감염자 본인은 제외
    """
    result = {}                         # 결과 저장: 이름 → (레벨, 경로)
    visited = {start}                   # 이미 감염 처리한 사람 (중복 감염 방지)
    queue = deque([(start, 0, [start])])  # (현재 사람, 현재 레벨, 여기까지 온 경로)

    while queue:                        # 큐가 빌 때까지 반복
        node, level, path = queue.popleft()   # 가장 먼저 들어온 사람을 꺼냄

        if level == max_level:          # 이미 최대 단계면 더 퍼뜨리지 않음
            continue

        # graph.get(node, []) : 그래프에 key로 없는 사람(접촉 정보 없음)은 빈 리스트로 처리
        for neighbor in graph.get(node, []):
            if neighbor not in visited:         # 처음 감염되는 사람만 처리
                visited.add(neighbor)
                new_path = path + [neighbor]    # 경로에 이 사람을 이어 붙임
                result[neighbor] = (level + 1, new_path)
                queue.append((neighbor, level + 1, new_path))  # 다음 물결 후보로 줄 세우기

    return result

# ── 3. 결과 출력 함수 ────────────────────────────────
def print_infection(graph, start, max_level=2):
    result = trace_infection(graph, start, max_level)
    print(f"[최초 감염자: {start} / {max_level}단계 이내]")
    if not result:
        print("  전염된 사람 없음")
    for level in range(1, max_level + 1):       # 레벨별로 묶어서 출력
        people = [(n, p) for n, (l, p) in result.items() if l == level]
        if people:
            print(f"  ▶ Level {level} ({len(people)}명)")
            for name, path in people:
                print(f"     - {name:<15} 경로: {' → '.join(path)}")
    print()

print_infection(graph, 'Yujin Kim')
```
